```python
def setup_driver():
    options = Options()
    # options.add_argument('--headless')
    options.add_argument('--disable-gpu')
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-blink-features=AutomationControlled')
    driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()), options=options)
    driver.set_window_size(random.randint(800, 1200), random.randint(600, 800))
    actions = ActionChains(driver)
    print("Driver set")
    
    # Accepting Cookies
    driver.get(URL)
    cookie_button = """//*[@id="CookiesConsent"]/div/div/form/div/button"""
    cookie_btn = driver.find_element(By.XPATH, cookie_button)
    random_delay(4, 6)
    try:
        cookie_btn.click()
        print("Cookie Accepted")
        random_delay(2, 3)
    except Exception as e:
        print(f'Could not click on cookie button {URL}: {e}')
    return driver, actions
```

```python
def convert_currency_string(currency_str):
    if '£' in currency_str and '.' in currency_str and ',' in currency_str:
        currency_str = str(float(currency_str.strip('£').replace('.', '').replace(',', '.'))*1.16).replace('.', ',')
    elif '£' in currency_str:
        currency_str = str(float(currency_str.strip('£').replace(',', '.'))*1).replace('.', ',')
    cleaned_str = currency_str.strip('€ £')
    
    # If there's a comma and no period, it's likely a decimal comma
    if ',' in cleaned_str and '.' not in cleaned_str:
        cleaned_str = cleaned_str.replace(',', '.')
    else:
        # Remove periods used as thousand separators
        cleaned_str = cleaned_str.replace('.', '').replace(',', '.')

    return float(cleaned_str)
```

```python
def get_df(data, category, selected_expansions = [], selected_cards = []):
    cards = []
    for expansion, card_list in data[category].items():
        for card_name, card_info in card_list.items():
            if card_info['quantity'] == 0:
                continue
            card_info_copy = card_info.copy()  # Avoid modifying the original data
            card_info_copy.update({'expansion': expansion, 'card_name': card_name})
            cards.append(card_info_copy)

    df = pd.DataFrame(cards).set_index(['expansion', 'card_name'])

    if selected_expansions or selected_cards:
        df = df.loc[df.index.get_level_values('expansion').isin(selected_expansions) | 
                    df.index.get_level_values('card_name').isin(selected_cards)].copy()
    
    df = df[df['quantity'] > 0]
    
    return df
```

```python
def retry_on_failure(retries=3, delay=2):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            last_exception = None
            for attempt in range(retries):
                try:
                    return func(*args, **kwargs)
                except (WebDriverException, InvalidSessionIdException) as e:
                    last_exception = e
                    print(f"Attempt {attempt + 1} failed. Retrying in {delay} seconds...")
                    time.sleep(delay)
                    # If it's a session issue, try to reinitialize the driver
                    if isinstance(e, InvalidSessionIdException):
                        try:
                            driver, actions = setup_driver()
                            # Update the global driver
                            globals()['driver'] = driver
                            globals()['actions'] = actions
                        except Exception as init_error:
                            print(f"Failed to reinitialize driver: {init_error}")
            raise last_exception
        return wrapper
    return decorator
```

```python
@retry_on_failure(retries=3, delay=2)
def get_card_price(url, driver, actions):
    """Get price information for a card with retry logic"""
    try:
        driver.get(url)
        random_delay(1, 3)
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        random_delay(3, 4)
        
        prices = {}
        
        try:
            p_from = convert_currency_string(driver.find_element(By.XPATH, price_from).text)
            prices['price_from'] = p_from
        except Exception as e:
            print(f'Could not retrieve price_from for {url}: {e}')
            
        try:
            p_trend = convert_currency_string(driver.find_element(By.XPATH, price_trend).text)
            prices['price_trend'] = p_trend
        except Exception as e:
            print(f'Could not retrieve price_trend for {url}: {e}')
            
        try:
            p_30_day = convert_currency_string(driver.find_element(By.XPATH, price_30_day_avg).text)
            prices['price_30_day_avg'] = p_30_day
        except Exception as e:
            print(f'Could not retrieve price_30_day_avg for {url}: {e}')
            
        return prices
            
    except Exception as e:
        print(f"Error getting price for {url}: {str(e)}")
        raise
```

```python
def get_price(data, category, selected_expansions = [], selected_cards = []):
    global driver, actions
    df = get_df(data, category, selected_expansions, selected_cards)

    counter = 0
    for index, row in df.iterrows():
        if counter == counter_threshold:
            save_data(data)
            save_updated_cards_prices(updated_cards_prices)
            driver.quit()
            driver, actions = setup_driver()
            counter = 0
            
        expansion, card_name = index
        if expansion in updated_cards_prices and card_name in updated_cards_prices[expansion]:
            continue
            
        print(f"Getting price for: {card_name}")
        url = f"{URL}/{category}/{expansion}/{card_name}"
        
        try:
            prices = get_card_price(url, driver, actions)
            
            # Update data and DataFrame with retrieved prices
            if 'price_from' in prices:
                df.at[index, 'price_from'] = prices['price_from']
                data[category][expansion][card_name]['price_from'] = prices['price_from']
                
            if 'price_trend' in prices:
                data[category][expansion][card_name]['price_trend'] = prices['price_trend']
                
            if 'price_30_day_avg' in prices:
                df.at[index, 'price_30_day_avg'] = prices['price_30_day_avg']
                data[category][expansion][card_name]['price_30_day_avg'] = prices['price_30_day_avg']

            if expansion in updated_cards_prices:
                updated_cards_prices[expansion].append(card_name)
            else:
                updated_cards_prices[expansion] = [card_name]
                
            counter += 1
            
        except Exception as e:
            print(f"Failed to process card {card_name}: {str(e)}")
            # Save progress even if there's an error
            save_data(data)
            save_updated_cards_prices(updated_cards_prices)
            
    save_data(data)
    save_updated_cards_prices(updated_cards_prices)
```

```python
def get_worth_google_sheets(df, spreadsheet_id):
    """
    Export dataframe to Google Sheets with formatting and hierarchical grouping
    """
    # Reset index to get expansion and card_name as columns
    df = df.reset_index()
    
    # Calculate totals for each card
    df['total_price_from'] = df['quantity'] * df['price_from']    
    df['total_price_trend'] = df['quantity'] * df['price_trend']    
    df['total_price_price_30_day_avg'] = df['quantity'] * df['price_30_day_avg']    
    
    # Create expansion subtotals
    expansion_totals = df.groupby('expansion').agg({
        'quantity': 'sum',
        'total_price_from': 'sum',
        'total_price_trend': 'sum',
        'total_price_price_30_day_avg': 'sum'
    }).reset_index()
    
    # Create hierarchical view
    rows = []
    for expansion in sorted(df['expansion'].unique()):
        # Add expansion row
        exp_total = expansion_totals[expansion_totals['expansion'] == expansion].iloc[0]
        rows.append({
            'expansion': expansion,
            'card_name': '',  # Empty for expansion rows
            'quantity': exp_total['quantity'],
            'price_from': '',  # Empty for expansion rows
            'price_trend': '',  # Empty for expansion rows
            'price_30_day_avg': '',  # Empty for expansion rows
            'total_price_from': exp_total['total_price_from'],
            'total_price_trend': exp_total['total_price_trend'],
            'total_price_price_30_day_avg': exp_total['total_price_price_30_day_avg']
        })
        
        # Add card rows for this expansion
        exp_cards = df[df['expansion'] == expansion].sort_values('card_name')
        for _, card in exp_cards.iterrows():
            rows.append({
                'expansion': '',  # Empty for card rows
                'card_name': '  ' + card['card_name'],  # Add indentation
                'quantity': card['quantity'],
                'price_from': card['price_from'],
                'price_trend': card['price_trend'],
                'price_30_day_avg': card['price_30_day_avg'],
                'total_price_from': card['total_price_from'],
                'total_price_trend': card['total_price_trend'],
                'total_price_price_30_day_avg': card['total_price_price_30_day_avg']
            })
    
    # Create DataFrame from rows
    export_df = pd.DataFrame(rows)
    
    # Add grand total row
    total_row = pd.DataFrame([{
        'expansion': 'Grand Total',
        'card_name': '',
        'quantity': df['quantity'].sum(),
        'price_from': '',  # Empty for total row
        'price_trend': '',  # Empty for total row
        'price_30_day_avg': '',  # Empty for total row
        'total_price_from': df['total_price_from'].sum(),
        'total_price_trend': df['total_price_trend'].sum(),
        'total_price_price_30_day_avg': df['total_price_price_30_day_avg'].sum()
    }])
    
    export_df = pd.concat([export_df, total_row])
    
    # Replace NaN values
    export_df = export_df.fillna('')
    
    # Rest of the code remains the same...
    try:
        SCOPES = ['https://www.googleapis.com/auth/spreadsheets']
        creds = service_account.Credentials.from_service_account_file(
            'credentials/card-market-pricer-44468fb06eaa.json',
            scopes=SCOPES
        )
        
        service = build('sheets', 'v4', credentials=creds)
        
        # First, check if the sheet exists and create it if it doesn't
        try:
            sheet_metadata = service.spreadsheets().get(spreadsheetId=spreadsheet_id).execute()
            sheet_exists = False
            for sheet in sheet_metadata.get('sheets', ''):
                if sheet.get('properties', {}).get('title') == 'Collection':
                    sheet_exists = True
                    sheet_id = sheet.get('properties', {}).get('sheetId')
                    break
            
            if not sheet_exists:
                request_body = {
                    'requests': [{
                        'addSheet': {
                            'properties': {
                                'title': 'Collection'
                            }
                        }
                    }]
                }
                response = service.spreadsheets().batchUpdate(
                    spreadsheetId=spreadsheet_id,
                    body=request_body
                ).execute()
                sheet_id = response['replies'][0]['addSheet']['properties']['sheetId']
        except Exception as e:
            print(f"Error checking/creating sheet: {e}")
            return
        
        # Convert DataFrame to values list
        values = [export_df.columns.tolist()]
        for _, row in export_df.iterrows():
            row_values = []
            for value in row:
                if pd.isna(value) or value == '':
                    row_values.append('')
                elif isinstance(value, (int, float)):
                    row_values.append(float(value))
                else:
                    row_values.append(str(value))
            values.append(row_values)
        
        # Clear existing content
        service.spreadsheets().values().clear(
            spreadsheetId=spreadsheet_id,
            range='Collection!A:Z'
        ).execute()
        
        # Update values
        body = {
            'values': values
        }
        service.spreadsheets().values().update(
            spreadsheetId=spreadsheet_id,
            range='Collection!A1',
            valueInputOption='RAW',
            body=body
        ).execute()
        
        # Apply formatting
        requests = [
            # Column widths
            {
                "updateDimensionProperties": {
                    "range": {
                        "sheetId": sheet_id,
                        "dimension": "COLUMNS",
                        "startIndex": 0,
                        "endIndex": 2
                    },
                    "properties": {
                        "pixelSize": 250  # Wider for expansion and card names
                    },
                    "fields": "pixelSize"
                }
            },
            {
                "updateDimensionProperties": {
                    "range": {
                        "sheetId": sheet_id,
                        "dimension": "COLUMNS",
                        "startIndex": 2,
                        "endIndex": len(export_df.columns)
                    },
                    "properties": {
                        "pixelSize": 120
                    },
                    "fields": "pixelSize"
                }
            },
            # Headers formatting
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": 0,
                        "endRowIndex": 1
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "backgroundColor": {"red": 0.8, "green": 0.8, "blue": 0.8},
                            "textFormat": {"bold": True},
                            "horizontalAlignment": "CENTER"
                        }
                    },
                    "fields": "userEnteredFormat(backgroundColor,textFormat,horizontalAlignment)"
                }
            },
            # Expansion rows formatting
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": 1,
                        "endRowIndex": len(values)
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "backgroundColor": {"red": 0.95, "green": 0.95, "blue": 0.95},
                            "textFormat": {"bold": True}
                        }
                    },
                    "fields": "userEnteredFormat(backgroundColor,textFormat)"
                }
            },
            # Currency formatting for individual prices
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": 1,
                        "endRowIndex": len(values),
                        "startColumnIndex": 3,  # Starting from price columns
                        "endColumnIndex": 6  # Up to total price columns
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "numberFormat": {
                                "type": "CURRENCY",
                                "pattern": "€ #,##0.00"
                            }
                        }
                    },
                    "fields": "userEnteredFormat.numberFormat"
                }
            },
            # Currency formatting for total prices
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": 1,
                        "endRowIndex": len(values),
                        "startColumnIndex": 6,  # Starting from total price columns
                        "endColumnIndex": len(export_df.columns)
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "numberFormat": {
                                "type": "CURRENCY",
                                "pattern": "€ #,##0.00"
                            }
                        }
                    },
                    "fields": "userEnteredFormat.numberFormat"
                }
            },
            # Grand total row formatting
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": len(values) - 1,
                        "endRowIndex": len(values),
                        "startColumnIndex": 3,
                        "endColumnIndex": len(export_df.columns)
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "backgroundColor": {"red": 1, "green": 0.9, "blue": 0.7},
                            "textFormat": {"bold": True},
                            "numberFormat": {
                                "type": "CURRENCY",
                                "pattern": "€ #,##0.00"
                            }
                        }
                    },
                    "fields": "userEnteredFormat(backgroundColor,textFormat,numberFormat)"
                }
            },
            # Grand total row formatting for non-currency columns (expansion, card_name, quantity)
            {
                "repeatCell": {
                    "range": {
                        "sheetId": sheet_id,
                        "startRowIndex": len(values) - 1,
                        "endRowIndex": len(values),
                        "startColumnIndex": 0,
                        "endColumnIndex": 3  # Up to but not including price columns
                    },
                    "cell": {
                        "userEnteredFormat": {
                            "backgroundColor": {"red": 1, "green": 0.9, "blue": 0.7},
                            "textFormat": {"bold": True},
                            "numberFormat": {
                                "type": "NUMBER",
                                "pattern": "#,##0"
                            }
                        }
                    },
                    "fields": "userEnteredFormat(backgroundColor,textFormat,numberFormat)"
                }
            },
            # Freeze panes
            {
                "updateSheetProperties": {
                    "properties": {
                        "sheetId": sheet_id,
                        "gridProperties": {
                            "frozenRowCount": 1,
                            "frozenColumnCount": 2
                        }
                    },
                    "fields": "gridProperties.frozenRowCount,gridProperties.frozenColumnCount"
                }
            }
        ]
        
        # Execute all formatting requests
        service.spreadsheets().batchUpdate(
            spreadsheetId=spreadsheet_id,
            body={'requests': requests}
        ).execute()
        
        print(f"Successfully updated Google Sheet")
        
    except HttpError as error:
        print(f"An error occurred: {error}")
        return error
```
